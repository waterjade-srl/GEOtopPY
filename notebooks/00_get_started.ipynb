# 00 · Get started

Get the code, install it, and run a short 1D reference case from the command
line, as you would run `geotop`.

Steps 1-3 are done in a terminal (Linux or macOS); the cells from step 4 on
run here.

## 1. Get the code

```bash
git clone https://github.com/geotopmodel/GEOtopPY.git
cd GEOtopPY
```

## 2. Install

Python 3.9 or newer:

```bash
python3 -m venv .venv
. .venv/bin/activate              # Windows: .venv\Scripts\activate
python -m pip install -e '.[notebooks]'
```

The model needs only the standard library; the `notebooks` extra adds
Jupyter, pandas and matplotlib.

## 3. Open this notebook

With the environment active:

```bash
jupyter lab notebooks/00_get_started.ipynb
```

## 4. Check the command

````python
!GEOtopPY --help
````

## 5. The case

`ColdelaPorte`: one point, one day (1 August 1993), hourly step, energy and
water balance. A standard GEOtop case directory; `output-tabs-SE27XX/` holds
the outputs of the official C++ run.

````python
!ls ../tests/1D/ColdelaPorte
````

## 6. Run it

The same call as `geotop <dir>`. It takes about a second.

````python
!GEOtopPY ../tests/1D/ColdelaPorte
````
